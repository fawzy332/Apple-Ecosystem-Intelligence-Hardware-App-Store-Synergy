```python
import pandas as pd

# Load the datasets into pandas DataFrames
AS_df = pd.read_csv('/content/apple_global_sales_dataset.csv')
APP_df = pd.read_csv('/content/appleAppData.csv')

# Display the first few rows of each DataFrame to verify
print("Apple Global Sales Dataset:")
display(AS_df.head())

print("\nApple App Data:")
display(APP_df.head())
```

```python
#Applying Mode to nulls in storage column to fix missing data
mode_storage=AS_df["storage"].mode()[0]
AS_df["storage"]=AS_df["storage"].fillna(mode_storage).replace(True)
```

```python
#Applying a standard text to fill null cells in previous software column
AS_df["previous_device_os"]=AS_df["previous_device_os"].fillna("No Previous Software")
```

```python
from numpy import median
#Inserting median in customer ratings column to fill nulls since non nulls is a bigger percentage than nulls
Median_rating=AS_df["customer_rating"].median()
AS_df["customer_rating"]=AS_df['customer_rating'].fillna(Median_rating).replace(True)
```

```python
# converting date column safely by telling pandas to automatically handle formatting, prioritizing day-first
AS_df['sale_date'] = pd.to_datetime(AS_df['sale_date'], format='mixed', dayfirst=True)
```

```python
#removing unnecessary columns from app dataset
APP_df=APP_df.drop(columns=['Developer_Url','Developer_Website'])
```

```python
#removing null app name row
APP_df=APP_df.dropna(subset=['App_Name'])
```

```python
#Replace null values in relased column with mode and change its type to date
modereleased=APP_df['Released'].mode()[0]
APP_df['Released']=APP_df['Released'].fillna(modereleased).replace(True)
APP_df['Released']= pd.to_datetime(APP_df['Released'], format='mixed', dayfirst=True)
```

```python
modeupdated=APP_df['Updated'].mode()[0]
APP_df['Updated']=APP_df['Updated'].fillna(modereleased).replace(True)
APP_df['Updated']= pd.to_datetime(APP_df['Updated'], format='mixed', dayfirst=True)
```

```python
#since a big percentage of prices in app dataset is the same (0.00) but rows with null price values have free columns set as false mean will be applied to nulls
```

```python
# Calculate the median price excluding free apps (Price == 0)
median_non_zero_price = APP_df[APP_df['Price'] > 0]['Price'].median()

# Fill the null values in the Price column with this calculated median
APP_df['Price'] = APP_df['Price'].fillna(median_non_zero_price)
```

```python
# Exporting both clean DataFrames to CSV files
AS_df.to_csv('cleaned_apple_global_sales_dataset.csv', index=False)
APP_df.to_csv('cleaned_apple_app_data.csv', index=False)
```
